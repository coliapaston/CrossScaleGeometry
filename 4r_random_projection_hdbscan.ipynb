# Random Orthogonal Projection and HDBSCAN Control

Generate one full orthogonal basis per model and seed, then use nested slices at the nine baseline dimensions.

```python
import gc
import sys

import torch

sys.path.append("./cluster_anlys")

from random_projection import (
    DEFAULT_SEEDS,
    MODEL_CONFIGS,
    aggregate_random_projection_seed_summaries,
    run_random_projection_hdbscan_grid,
)
```

```python
OUT_ROOT = "comp"
SPACE_NAME = "output_proj"
SEED_LIST = list(DEFAULT_SEEDS)
ROW_BATCH_SIZE = 4096

HDBSCAN_CONFIG = {
    "l2_norm": True,
    "min_cluster_size": 5,
    "min_samples": 5,
    "metric": "euclidean",
    "cluster_selection_method": "eom",
    "cluster_selection_epsilon": 0.0,
}
```

```python
def run_model_random_projection_control(model_key):
    model_config = MODEL_CONFIGS[model_key]
    model_name = model_config["model_name"]
    candidate_dims = model_config["candidate_dims"]
    tensor_path = f"{model_name}/tensors/{SPACE_NAME}.pt"

    embedding_matrix = torch.load(tensor_path, map_location="cpu")
    embedding_matrix = embedding_matrix.to(torch.float32)
    print(f"Loaded {model_name}: {tuple(embedding_matrix.shape)}")

    seed_summaries = {}
    for rp_seed in SEED_LIST:
        seed_summaries[int(rp_seed)] = run_random_projection_hdbscan_grid(
            embedding_matrix=embedding_matrix,
            candidate_dims=candidate_dims,
            model_name=model_name,
            space_name=SPACE_NAME,
            rp_seed=int(rp_seed),
            out_root=OUT_ROOT,
            row_batch_size=ROW_BATCH_SIZE,
            reuse_complete=True,
            **HDBSCAN_CONFIG,
        )

    all_seed_summary_df = aggregate_random_projection_seed_summaries(
        out_root=OUT_ROOT,
        model_name=model_name,
        space_name=SPACE_NAME,
        seed_list=SEED_LIST,
    )

    del embedding_matrix
    gc.collect()
    return seed_summaries, all_seed_summary_df
```

```python
from pathlib import Path

import numpy as np
import pandas as pd
from IPython.display import display


def canonicalize_cluster_partition(token_ids, cluster_ids):
    token_ids = np.asarray(token_ids, dtype=np.int64)
    cluster_ids = np.asarray(cluster_ids, dtype=np.int64)
    canonical_ids = np.full(cluster_ids.shape, -1, dtype=np.int64)
    assigned = cluster_ids >= 0
    if assigned.any():
        assigned_tokens = pd.Series(token_ids[assigned])
        assigned_clusters = pd.Series(cluster_ids[assigned])
        canonical_ids[assigned] = (
            assigned_tokens.groupby(assigned_clusters, sort=False).transform("min")
        ).to_numpy(dtype=np.int64)
    return canonical_ids


def check_full_dimension_hdbscan_invariance(model_key):
    model_config = MODEL_CONFIGS[model_key]
    model_name = model_config["model_name"]
    full_dimension = int(max(model_config["candidate_dims"]))
    model_root = Path(OUT_ROOT) / model_name / SPACE_NAME
    baseline_summary_path = model_root / "summary.csv"
    baseline_summary = pd.read_csv(baseline_summary_path)
    baseline_rows = baseline_summary.loc[
        baseline_summary["pca_dim"].astype(int) == full_dimension
    ]
    if len(baseline_rows) != 1:
        raise ValueError(
            f"Expected one baseline row at dim={full_dimension}, "
            f"found {len(baseline_rows)}"
        )

    baseline_row = baseline_rows.iloc[0]
    baseline_clusters = pd.read_csv(
        Path(baseline_row["cluster_csv_path"]),
        usecols=["token_id", "cluster_id", "probability"],
    ).sort_values("token_id").reset_index(drop=True)
    if baseline_clusters["token_id"].duplicated().any():
        raise ValueError("Baseline cluster output contains duplicate token IDs")

    baseline_token_ids = baseline_clusters["token_id"].to_numpy(dtype=np.int64)
    baseline_cluster_ids = baseline_clusters["cluster_id"].to_numpy(dtype=np.int64)
    baseline_probabilities = baseline_clusters["probability"].to_numpy(dtype=np.float64)
    baseline_canonical_ids = canonicalize_cluster_partition(
        baseline_token_ids, baseline_cluster_ids
    )

    records = []
    for rp_seed in SEED_LIST:
        seed_summary_path = (
            model_root
            / "random_projection"
            / f"seed_{int(rp_seed)}"
            / "summary.csv"
        )
        if not seed_summary_path.exists():
            records.append(
                {
                    "model_name": model_name,
                    "rp_seed": int(rp_seed),
                    "full_dimension": full_dimension,
                    "status": "seed_summary_missing",
                }
            )
            continue

        seed_summary = pd.read_csv(seed_summary_path)
        seed_rows = seed_summary.loc[
            seed_summary["pca_dim"].astype(int) == full_dimension
        ]
        if len(seed_rows) != 1:
            records.append(
                {
                    "model_name": model_name,
                    "rp_seed": int(rp_seed),
                    "full_dimension": full_dimension,
                    "status": "full_dimension_result_missing",
                }
            )
            continue

        seed_row = seed_rows.iloc[0]
        projected_clusters = pd.read_csv(
            Path(seed_row["cluster_csv_path"]),
            usecols=["token_id", "cluster_id", "probability"],
        ).sort_values("token_id").reset_index(drop=True)
        if projected_clusters["token_id"].duplicated().any():
            raise ValueError(
                f"Random-projection output contains duplicate token IDs for seed={rp_seed}"
            )

        projected_token_ids = projected_clusters["token_id"].to_numpy(dtype=np.int64)
        token_ids_exact = np.array_equal(baseline_token_ids, projected_token_ids)
        if not token_ids_exact:
            records.append(
                {
                    "model_name": model_name,
                    "rp_seed": int(rp_seed),
                    "full_dimension": full_dimension,
                    "status": "token_ids_mismatch",
                    "token_ids_exact": False,
                }
            )
            continue

        projected_cluster_ids = projected_clusters["cluster_id"].to_numpy(
            dtype=np.int64
        )
        projected_probabilities = projected_clusters["probability"].to_numpy(
            dtype=np.float64
        )
        projected_canonical_ids = canonicalize_cluster_partition(
            projected_token_ids, projected_cluster_ids
        )
        partition_mismatch_count = int(
            np.count_nonzero(baseline_canonical_ids != projected_canonical_ids)
        )
        noise_mismatch_count = int(
            np.count_nonzero(
                (baseline_cluster_ids == -1) != (projected_cluster_ids == -1)
            )
        )
        probability_difference = np.abs(
            baseline_probabilities - projected_probabilities
        )
        records.append(
            {
                "model_name": model_name,
                "rp_seed": int(rp_seed),
                "full_dimension": full_dimension,
                "status": "checked",
                "token_ids_exact": True,
                "partition_exact_up_to_label_permutation": (
                    partition_mismatch_count == 0
                ),
                "partition_mismatch_count": partition_mismatch_count,
                "noise_assignment_mismatch_count": noise_mismatch_count,
                "baseline_cluster_count": int(
                    baseline_row["n_clusters_excl_noise"]
                ),
                "projected_cluster_count": int(
                    seed_row["n_clusters_excl_noise"]
                ),
                "baseline_noise_count": int(baseline_row["n_noise"]),
                "projected_noise_count": int(seed_row["n_noise"]),
                "probabilities_exact": np.array_equal(
                    baseline_probabilities,
                    projected_probabilities,
                    equal_nan=True,
                ),
                "max_probability_abs_difference": float(
                    np.nanmax(probability_difference)
                ),
            }
        )

    result = pd.DataFrame(records)
    output_path = (
        model_root
        / "random_projection"
        / "full_dimension_invariance_check.csv"
    )
    output_path.parent.mkdir(parents=True, exist_ok=True)
    result.to_csv(output_path, index=False)
    checked = result.loc[result["status"] == "checked"]
    all_partitions_exact = bool(
        len(checked) == len(SEED_LIST)
        and checked["partition_exact_up_to_label_permutation"].all()
    )
    print(f"{model_name}: all full-dimensional partitions exact = {all_partitions_exact}")
    print(f"Saved invariance check: {output_path}")
    return result
```

```python
from sklearn.metrics import adjusted_mutual_info_score, adjusted_rand_score


def calculate_full_dimension_ami_ari(model_key):
    model_config = MODEL_CONFIGS[model_key]
    model_name = model_config["model_name"]
    full_dimension = int(max(model_config["candidate_dims"]))
    model_root = Path(OUT_ROOT) / model_name / SPACE_NAME

    baseline_summary = pd.read_csv(model_root / "summary.csv")
    baseline_rows = baseline_summary.loc[
        baseline_summary["pca_dim"].astype(int) == full_dimension
    ]
    if len(baseline_rows) != 1:
        raise ValueError(
            f"Expected one baseline row at dim={full_dimension}, "
            f"found {len(baseline_rows)}"
        )

    baseline_clusters = pd.read_csv(
        Path(baseline_rows.iloc[0]["cluster_csv_path"]),
        usecols=["token_id", "cluster_id"],
    ).sort_values("token_id").reset_index(drop=True)
    baseline_token_ids = baseline_clusters["token_id"].to_numpy(dtype=np.int64)
    baseline_labels = baseline_clusters["cluster_id"].to_numpy(dtype=np.int64)

    records = []
    for rp_seed in SEED_LIST:
        seed_summary_path = (
            model_root
            / "random_projection"
            / f"seed_{int(rp_seed)}"
            / "summary.csv"
        )
        if not seed_summary_path.exists():
            records.append(
                {
                    "model_name": model_name,
                    "rp_seed": int(rp_seed),
                    "full_dimension": full_dimension,
                    "status": "seed_summary_missing",
                }
            )
            continue

        seed_summary = pd.read_csv(seed_summary_path)
        seed_rows = seed_summary.loc[
            seed_summary["pca_dim"].astype(int) == full_dimension
        ]
        if len(seed_rows) != 1:
            records.append(
                {
                    "model_name": model_name,
                    "rp_seed": int(rp_seed),
                    "full_dimension": full_dimension,
                    "status": "full_dimension_result_missing",
                }
            )
            continue

        projected_clusters = pd.read_csv(
            Path(seed_rows.iloc[0]["cluster_csv_path"]),
            usecols=["token_id", "cluster_id"],
        ).sort_values("token_id").reset_index(drop=True)
        projected_token_ids = projected_clusters["token_id"].to_numpy(
            dtype=np.int64
        )
        if not np.array_equal(baseline_token_ids, projected_token_ids):
            records.append(
                {
                    "model_name": model_name,
                    "rp_seed": int(rp_seed),
                    "full_dimension": full_dimension,
                    "status": "token_ids_mismatch",
                }
            )
            continue

        projected_labels = projected_clusters["cluster_id"].to_numpy(
            dtype=np.int64
        )
        records.append(
            {
                "model_name": model_name,
                "rp_seed": int(rp_seed),
                "full_dimension": full_dimension,
                "status": "checked",
                "n_tokens": int(len(baseline_labels)),
                "ari": float(
                    adjusted_rand_score(baseline_labels, projected_labels)
                ),
                "ami": float(
                    adjusted_mutual_info_score(
                        baseline_labels,
                        projected_labels,
                        average_method="arithmetic",
                    )
                ),
            }
        )

    result = pd.DataFrame(records)
    output_path = (
        model_root
        / "random_projection"
        / "full_dimension_ami_ari.csv"
    )
    output_path.parent.mkdir(parents=True, exist_ok=True)
    result.to_csv(output_path, index=False)
    print(f"Saved full-dimensional AMI/ARI check: {output_path}")
    return result
```
