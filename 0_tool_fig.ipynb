```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

model_specs = [
    {"model_name": "gpt-oss", "space_name": "output_proj", "label": "gpt-oss"},
    {"model_name": "mistralai/Mistral-7B-v0.1", "space_name": "output_proj", "label": "Mistral-7B"},
    {"model_name": "mistralai/Mixtral-8x7B-v0.1", "space_name": "output_proj", "label": "Mixtral-8x7B"},
]

out_root = "comp"

# Adjustable style parameters for the two final plots below.
FIGSIZE_W = 8
FIGSIZE_H = 5
TITLE_FONTSIZE = 16
LABEL_FONTSIZE = 13
TICK_FONTSIZE = 11
LEGEND_FONTSIZE = 10

full_partition_stats = {}

for spec in model_specs:
    summary_path = Path(out_root) / spec["model_name"] / spec["space_name"] / "summary.csv"
    summary_df = pd.read_csv(summary_path).copy()
    summary_df["run_id"] = pd.to_numeric(summary_df["run_id"], errors="raise").astype(int)
    summary_df["n_clusters_excl_noise"] = pd.to_numeric(summary_df["n_clusters_excl_noise"], errors="raise").astype(int)
    summary_df = summary_df.sort_values("run_id").reset_index(drop=True)

    cluster_size_rows = []
    for _, row in summary_df.iterrows():
        cluster_df = pd.read_csv(row["cluster_csv_path"]).copy()
        cluster_df["cluster_id"] = pd.to_numeric(cluster_df["cluster_id"], errors="raise").astype(int)
        cluster_df = cluster_df.loc[cluster_df["cluster_id"] != -1].copy()
        cluster_sizes = cluster_df.groupby("cluster_id").size().to_numpy(dtype=float)
        cluster_size_rows.append({
            "run_id": int(row["run_id"]),
            "cluster_size_mean": float(np.mean(cluster_sizes)),
            "cluster_size_median": float(np.median(cluster_sizes)),
            "cluster_size_q25": float(np.quantile(cluster_sizes, 0.25)),
            "cluster_size_q75": float(np.quantile(cluster_sizes, 0.75)),
        })

    cluster_size_df = pd.DataFrame(cluster_size_rows).sort_values("run_id").reset_index(drop=True)
    merged_df = summary_df[["run_id", "n_clusters_excl_noise"]].merge(cluster_size_df, on="run_id", how="inner")
    full_partition_stats[spec["label"]] = merged_df

full_partition_stats["gpt-oss"].head()
```

# Full PCA vs Random Orthogonal Projection Macro-Trend Figures

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
```

```python
from typing import Any, Dict, Sequence

MODEL_CONFIGS: Dict[str, Dict[str, Any]] = {
    "mistral": {
        "model_name": "mistralai/Mistral-7B-v0.1",
        "tokenizer_path": "mistralai/Mistral-7B-v0.1/tokenizer",
        "candidate_dims": [5, 142, 997, 2084, 3031, 3459, 3938, 4088, 4096],
    },
    "mixtral": {
        "model_name": "mistralai/Mixtral-8x7B-v0.1",
        "tokenizer_path": "mistralai/Mixtral-8x7B-v0.1/tokenizer",
        "candidate_dims": [8, 158, 1111, 2156, 3052, 3457, 3957, 4093, 4096],
    },
    "gpt-oss": {
        "model_name": "gpt-oss",
        "tokenizer_path": "gpt-oss/tokenizer",
        "candidate_dims": [6, 182, 466, 739, 1591, 2264, 2532, 2868, 2880],
    },
}
```

```python
def plot_pca_vs_random_projection(
    model_key,
    *,
    out_root="comp",
    space_name="output_proj",
    save_dir="res/linguistic",
):
    model_name = MODEL_CONFIGS[model_key]["model_name"]
    model_dir = Path(out_root) / model_name / space_name
    rp_dir = model_dir / "random_projection"
    baseline_morph = pd.read_csv(model_dir / "kondrak_global_summary.csv")
    baseline_script = pd.read_csv(model_dir / "script_entropy_summary.csv")
    rp_morph = pd.read_csv(rp_dir / "kondrak_global_summary_all_seeds.csv")
    rp_script = pd.read_csv(rp_dir / "script_entropy_summary_all_seeds.csv")

    rp_morph_curve = (
        rp_morph.groupby("pca_dim", as_index=False)
        .agg(rp_mean=("global_mean", "mean"), rp_sd=("global_mean", "std"))
        .sort_values("pca_dim")
    )
    rp_script_curve = (
        rp_script.groupby("pca_dim", as_index=False)
        .agg(rp_mean=("mean_H", "mean"), rp_sd=("mean_H", "std"))
        .sort_values("pca_dim")
    )

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    panels = [
        (axes[0], baseline_morph, rp_morph_curve, "global_mean", "Orthographic Similarity"),
        (axes[1], baseline_script, rp_script_curve, "mean_H", "Multi-script Entropy"),
    ]
    for axis, baseline_df, rp_curve, value_col, ylabel in panels:
        baseline_df = baseline_df.sort_values("pca_dim")
        x = rp_curve["pca_dim"].to_numpy(dtype=int)
        rp_mean = rp_curve["rp_mean"].to_numpy(dtype=float)
        rp_sd = rp_curve["rp_sd"].fillna(0.0).to_numpy(dtype=float)
        axis.plot(
            baseline_df["pca_dim"],
            baseline_df[value_col],
            marker="o",
            linewidth=2,
            label="Full PCA",
        )
        axis.plot(x, rp_mean, marker="s", linewidth=2, label="Random projection")
        axis.fill_between(x, rp_mean - rp_sd, rp_mean + rp_sd, alpha=0.2)
        axis.set_xlabel("Dimension")
        axis.set_ylabel(ylabel)
        axis.legend()

    fig.suptitle(model_name)
    fig.tight_layout()
    save_dir = Path(save_dir)
    save_dir.mkdir(parents=True, exist_ok=True)
    safe_model_name = model_name.replace("/", "_")
    save_path = save_dir / f"random_projection_{safe_model_name}.pdf"
    fig.savefig(save_path)
    print(f"Saved to: {save_path}")
    plt.show()
    return save_path


random_projection_figure_paths = {
    model_key: plot_pca_vs_random_projection(model_key)
    for model_key in ["mistral", "mixtral", "gpt-oss"]
}
random_projection_figure_paths
```
