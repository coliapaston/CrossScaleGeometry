```python
import pandas as pd
from pathlib import Path
```

```python
out_root = "comp"

model_specs = [
    {"model_name": "gpt-oss", "space_name": "output_proj", "label": "gpt-oss"},
    {"model_name": "mistralai/Mistral-7B-v0.1", "space_name": "output_proj", "label": "Mistral-7B"},
    {"model_name": "mistralai/Mixtral-8x7B-v0.1", "space_name": "output_proj", "label": "Mixtral-8x7B"},
]
```

```python
summary_tables = {}

for spec in model_specs:
    summary_path = Path(out_root) / spec["model_name"] / spec["space_name"] / "summary.csv"
    df = pd.read_csv(summary_path).copy()
    df["run_id"] = pd.to_numeric(df["run_id"], errors="raise").astype(int)
    df["noise_rate"] = pd.to_numeric(df["noise_ratio"], errors="raise").astype(float)
    df["valid_token_rate"] = 1.0 - df["noise_rate"]
    df = df.sort_values("run_id").reset_index(drop=True)
    summary_tables[spec["label"]] = df[["run_id", "pca_dim", "noise_rate", "valid_token_rate", "n_noise", "n_total", "n_clusters_excl_noise"]]

summary_tables["gpt-oss"].head()
```
