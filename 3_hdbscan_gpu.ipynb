```python
import time
import numpy as np
import torch

def benchmark_gpu_hdbscan_rounds(
    embedding_matrix,
    n_rounds=5,
    min_cluster_size=5,
    min_samples=5,
    metric="euclidean",
    cluster_selection_method="eom",
    cluster_selection_epsilon=0.0,
    cleanup_between_rounds=True,
):
    if isinstance(embedding_matrix, torch.Tensor):
        X_np = embedding_matrix.detach().cpu().to(torch.float32).numpy()
    else:
        X_np = np.asarray(embedding_matrix, dtype=np.float32)

    if X_np.ndim != 2:
        raise ValueError(f"embedding_matrix must be 2D, got shape={X_np.shape}")

    X_np = np.asarray(X_np, dtype=np.float32)
    per_round_seconds = []

    if cleanup_between_rounds:
        cleanup_gpu_memory()

    total_start = time.perf_counter()
    for round_idx in range(int(n_rounds)):
        round_start = time.perf_counter()
        labels_np, probs_np = run_gpu_hdbscan(
            X_np,
            min_cluster_size=min_cluster_size,
            min_samples=min_samples,
            cluster_selection_method=cluster_selection_method,
            metric=metric,
            cluster_selection_epsilon=cluster_selection_epsilon,
        )
        round_elapsed = time.perf_counter() - round_start
        per_round_seconds.append(round_elapsed)

        del labels_np, probs_np
        if cleanup_between_rounds:
            cleanup_gpu_memory()

        print(f"Round {round_idx + 1}/{n_rounds}: {round_elapsed:.4f}s")

    total_elapsed = time.perf_counter() - total_start
    print(f"Total elapsed: {total_elapsed:.4f}s")
    print(f"Mean per round: {np.mean(per_round_seconds):.4f}s")

    return {
        "n_rounds": int(n_rounds),
        "total_seconds": float(total_elapsed),
        "mean_seconds": float(np.mean(per_round_seconds)),
        "per_round_seconds": per_round_seconds,
    }

# Example:
# timing_result = benchmark_gpu_hdbscan_rounds(embedding_matrix=my_embedding_matrix, n_rounds=10)
```
