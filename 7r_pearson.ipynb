# gpt-oss

# Permutation Vs Full


```python
# Compare each permutation-control model against the original full-PCA baseline.
comparison_specs = [
    # {
    #     "baseline_model_name": "gpt-oss",
    #     "perm_model_name": "permutation/gpt-oss_rand",
    #     "space_name": "output_proj",
    # },
    {
        "baseline_model_name": "mistralai/Mistral-7B-v0.1",
        "perm_model_name": "permutation/mistralai/Mistral-7B-v0.1_rand",
        "space_name": "output_proj",
    },
    {
        "baseline_model_name": "mistralai/Mixtral-8x7B-v0.1",
        "perm_model_name": "permutation/mistralai/Mixtral-8x7B-v0.1_rand",
        "space_name": "output_proj",
    },
]
```

```python
perm_vs_full_rows = []

for spec in comparison_specs:
    baseline_morph = standardize_morphology_summary_df(
        load_full_morphology_summary(
            out_root="comp",
            model_name=spec["baseline_model_name"],
            space_name=spec["space_name"],
        ),
        df_name="baseline_morphology_summary",
    )
    perm_morph = standardize_morphology_summary_df(
        load_full_morphology_summary(
            out_root="comp",
            model_name=spec["perm_model_name"],
            space_name=spec["space_name"],
            filename="kondrak_global_summary_mean.csv",
        ),
        df_name="permutation_morphology_summary",
    )
    baseline_script = standardize_script_summary_df(
        load_full_script_summary(
            out_root="comp",
            model_name=spec["baseline_model_name"],
            space_name=spec["space_name"],
        ),
        df_name="baseline_script_summary",
    )
    perm_script = standardize_script_summary_df(
        load_full_script_summary(
            out_root="comp",
            model_name=spec["perm_model_name"],
            space_name=spec["space_name"],
            filename="script_entropy_summary_mean.csv",
        ),
        df_name="permutation_script_summary",
    )

    morph_result = compare_seed_to_baseline_for_metric(
        baseline_morph,
        perm_morph,
        metric_family="morph",
        seed=0,
        n_permutations=5000,
        random_state=0,
    )
    script_result = compare_seed_to_baseline_for_metric(
        baseline_script,
        perm_script,
        metric_family="script",
        seed=0,
        n_permutations=5000,
        random_state=0,
    )
    joint_result = compare_seed_to_baseline_joint_mean(
        baseline_morph,
        perm_morph,
        baseline_script,
        perm_script,
        seed=0,
        n_permutations=5000,
        random_state=0,
    )

    perm_vs_full_rows.append({
        "baseline_model_name": spec["baseline_model_name"],
        "perm_model_name": spec["perm_model_name"],
        "morph_mean_r": morph_result["mean_r"],
        "morph_mean_p": morph_result["mean_p"],
        "script_mean_r": script_result["mean_r"],
        "script_mean_p": script_result["mean_p"],
        "joint_mean_r": joint_result["joint_mean_r"],
        "joint_mean_perm_p": joint_result["joint_mean_perm_p"],
        "morph_std_r": morph_result["std_r"],
        "morph_std_p": morph_result["std_p"],
        "script_std_r": script_result["std_r"],
        "script_std_p": script_result["std_p"],
        "morph_mean_std_joint_r": morph_result["mean_std_joint_r"],
        "morph_mean_std_joint_perm_p": morph_result["mean_std_joint_perm_p"],
        "script_mean_std_joint_r": script_result["mean_std_joint_r"],
        "script_mean_std_joint_perm_p": script_result["mean_std_joint_perm_p"],
    })

perm_vs_full_df = pd.DataFrame(perm_vs_full_rows)
perm_vs_full_df
```

```python
# Primary view for the permutation-control comparison.
perm_vs_full_df[[
    "baseline_model_name",
    "morph_mean_r",
    "morph_mean_p",
    "script_mean_r",
    "script_mean_p",
    "joint_mean_r",
    "joint_mean_perm_p",
]]
```

```python
# LaTeX table for the permutation-vs-full comparison.
perm_vs_full_latex = make_permutation_vs_full_latex_table(
    perm_vs_full_df,
    precision=3,
    model_order=[
        "gpt-oss",
        "mistralai/Mistral-7B-v0.1",
        "mistralai/Mixtral-8x7B-v0.1",
    ],
    caption="Pearson comparison between the original full-PCA results and the permutation control.",
    label="tab:perm_vs_full_pearson",
)

print(perm_vs_full_latex)
```
