# Length-bucket token counts


```python
from partition_permutation_control import filter_valid_cluster_df, load_cluster_df, load_summary_df
from permutation_multiseed_pipeline import (
    MODEL_CONFIGS,
    attach_length_bucket_column,
    load_tokenizer_for_model_config,
)

LENGTH_BUCKET_COLUMNS = ["1-2", "3-5", "6-10", "11+"]


def summarize_length_bucket_counts_for_model_key(
    model_key,
    *,
    out_root="comp",
    space_name="output_proj",
):
    model_config = MODEL_CONFIGS[model_key]
    source_model_name = model_config["source_model_name"]
    tokenizer = load_tokenizer_for_model_config(model_config)
    summary_df = load_summary_df(
        out_root=out_root,
        model_name=source_model_name,
        space_name=space_name,
    )
    token_text_cache = {}
    rows = []

    for _, row in summary_df.iterrows():
        cluster_df = load_cluster_df(row["cluster_csv_path"])
        valid_cluster_df = filter_valid_cluster_df(cluster_df)
        bucketed_df = attach_length_bucket_column(
            valid_cluster_df,
            tokenizer=tokenizer,
            model_name=source_model_name,
            token_text_cache=token_text_cache,
        )
        bucket_counts = bucketed_df["length_bucket"].value_counts().to_dict()
        out_row = {
            "model_key": model_key,
            "source_model_name": source_model_name,
            "run_id": int(row["run_id"]),
            "pca_dim": int(row["pca_dim"]),
            "n_non_noise": int(len(bucketed_df)),
        }
        for bucket_name in LENGTH_BUCKET_COLUMNS:
            safe_name = bucket_name.replace("+", "_plus").replace("-", "_")
            out_row[f"count_{safe_name}"] = int(bucket_counts.get(bucket_name, 0))
        rows.append(out_row)

    return pd.DataFrame(rows).sort_values("pca_dim").reset_index(drop=True)
```

```python
gpt_oss_length_bucket_counts_df = summarize_length_bucket_counts_for_model_key("gpt-oss")
mistral_length_bucket_counts_df = summarize_length_bucket_counts_for_model_key("mistral")
mixtral_length_bucket_counts_df = summarize_length_bucket_counts_for_model_key("mixtral")

length_bucket_counts_df = pd.concat(
    [
        gpt_oss_length_bucket_counts_df,
        mistral_length_bucket_counts_df,
        mixtral_length_bucket_counts_df,
    ],
    axis=0,
    ignore_index=True,
)

length_bucket_counts_outpath = Path("comp") / "permutation" / "length_bucket_token_counts.csv"
length_bucket_counts_outpath.parent.mkdir(parents=True, exist_ok=True)
length_bucket_counts_df.to_csv(length_bucket_counts_outpath, index=False)
print(length_bucket_counts_outpath)
length_bucket_counts_df
```
