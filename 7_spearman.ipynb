# Permutation Vs Full


```python
# Compare each permutation-control model against the original full-PCA baseline.
comparison_specs = [
    {
        "baseline_model_name": "gpt-oss",
        "perm_model_name": "permutation/gpt-oss_rand",
        "space_name": "output_proj",
    },
    {
        "baseline_model_name": "mistralai/Mistral-7B-v0.1",
        "perm_model_name": "permutation/mistralai/Mistral-7B-v0.1_rand",
        "space_name": "output_proj",
    },
    {
        "baseline_model_name": "mistralai/Mixtral-8x7B-v0.1",
        "perm_model_name": "permutation/mistralai/Mixtral-8x7B-v0.1_rand",
        "space_name": "output_proj",
    },
]
```

```python
perm_vs_full_rows = []

for spec in comparison_specs:
    baseline_morph = standardize_morphology_summary_df(
        load_full_morphology_summary(
            out_root="comp",
            model_name=spec["baseline_model_name"],
            space_name=spec["space_name"],
        ),
        df_name="baseline_morphology_summary",
    )
    perm_morph = standardize_morphology_summary_df(
        load_full_morphology_summary(
            out_root="comp",
            model_name=spec["perm_model_name"],
            space_name=spec["space_name"],
            filename="kondrak_global_summary_mean.csv",
        ),
        df_name="permutation_morphology_summary",
    )
    baseline_script = standardize_script_summary_df(
        load_full_script_summary(
            out_root="comp",
            model_name=spec["baseline_model_name"],
            space_name=spec["space_name"],
        ),
        df_name="baseline_script_summary",
    )
    perm_script = standardize_script_summary_df(
        load_full_script_summary(
            out_root="comp",
            model_name=spec["perm_model_name"],
            space_name=spec["space_name"],
            filename="script_entropy_summary_mean.csv",
        ),
        df_name="permutation_script_summary",
    )

    morph_result = compare_seed_to_baseline_for_metric(
        baseline_morph,
        perm_morph,
        metric_family="morph",
        seed=0,
        n_permutations=5000,
        random_state=0,
    )
    script_result = compare_seed_to_baseline_for_metric(
        baseline_script,
        perm_script,
        metric_family="script",
        seed=0,
        n_permutations=5000,
        random_state=0,
    )
    joint_result = compare_seed_to_baseline_joint_mean(
        baseline_morph,
        perm_morph,
        baseline_script,
        perm_script,
        seed=0,
        n_permutations=5000,
        random_state=0,
    )

    perm_vs_full_rows.append({
        "baseline_model_name": spec["baseline_model_name"],
        "perm_model_name": spec["perm_model_name"],
        "morph_mean_rho": morph_result["mean_rho"],
        "morph_mean_p": morph_result["mean_p"],
        "script_mean_rho": script_result["mean_rho"],
        "script_mean_p": script_result["mean_p"],
        "joint_mean_rho": joint_result["joint_mean_rho"],
        "joint_mean_perm_p": joint_result["joint_mean_perm_p"],
        "morph_std_rho": morph_result["std_rho"],
        "morph_std_p": morph_result["std_p"],
        "script_std_rho": script_result["std_rho"],
        "script_std_p": script_result["std_p"],
        "morph_mean_std_joint_rho": morph_result["mean_std_joint_rho"],
        "morph_mean_std_joint_perm_p": morph_result["mean_std_joint_perm_p"],
        "script_mean_std_joint_rho": script_result["mean_std_joint_rho"],
        "script_mean_std_joint_perm_p": script_result["mean_std_joint_perm_p"],
    })

perm_vs_full_df = pd.DataFrame(perm_vs_full_rows)
perm_vs_full_df
```

```python
def compute_length_bucket_morph_vs_full_spearman(
    comparison_specs,
    *,
    out_root="comp",
    n_permutations=5000,
    random_state=0,
):
    rows = []

    for spec in comparison_specs:
        baseline_morph = standardize_morphology_summary_df(
            load_full_morphology_summary(
                out_root=out_root,
                model_name=spec["baseline_model_name"],
                space_name=spec["space_name"],
            ),
            df_name="baseline_morphology_summary",
        )
        length_bucket_morph = standardize_morphology_summary_df(
            load_full_morphology_summary(
                out_root=out_root,
                model_name=spec["perm_model_name"],
                space_name=spec["space_name"],
                filename="kondrak_global_summary_mean.csv",
            ),
            df_name="length_bucket_morphology_summary",
        )

        morph_result = compare_seed_to_baseline_for_metric(
            baseline_morph,
            length_bucket_morph,
            metric_family="morph",
            seed=0,
            n_permutations=n_permutations,
            random_state=random_state,
        )

        rows.append({
            "baseline_model_name": spec["baseline_model_name"],
            "perm_model_name": spec["perm_model_name"],
            "n_dims": morph_result["n_dims"],
            "morph_mean_rho": morph_result["mean_rho"],
            "morph_mean_p": morph_result["mean_p"],
            "morph_std_rho": morph_result["std_rho"],
            "morph_std_p": morph_result["std_p"],
            "morph_mean_std_joint_rho": morph_result["mean_std_joint_rho"],
            "morph_mean_std_joint_perm_p": morph_result["mean_std_joint_perm_p"],
            "n_permutations": morph_result["mean_std_joint_n_permutations"],
        })

    return pd.DataFrame(rows)
```

# Length-bucketed permutation per-seed morphology Spearman


```python
def compute_length_bucket_seed_morph_vs_full_spearman(
    comparison_specs,
    *,
    seed_list,
    out_root="comp",
    n_permutations=5000,
    random_state=0,
):
    rows = []

    for spec in comparison_specs:
        baseline_morph = standardize_morphology_summary_df(
            load_full_morphology_summary(
                out_root=out_root,
                model_name=spec["baseline_model_name"],
                space_name=spec["space_name"],
            ),
            df_name="baseline_morphology_summary",
        )

        for perm_seed in seed_list:
            seed_morph = standardize_morphology_summary_df(
                load_full_morphology_summary(
                    out_root=out_root,
                    model_name=spec["perm_model_name"],
                    space_name=spec["space_name"],
                    filename=f"seeds/seed_{int(perm_seed)}/kondrak_global_summary.csv",
                ),
                df_name="length_bucket_seed_morphology_summary",
            )

            morph_result = compare_seed_to_baseline_for_metric(
                baseline_morph,
                seed_morph,
                metric_family="morph",
                seed=int(perm_seed),
                n_permutations=n_permutations,
                random_state=random_state,
            )

            rows.append({
                "baseline_model_name": spec["baseline_model_name"],
                "perm_model_name": spec["perm_model_name"],
                "perm_seed": int(perm_seed),
                "n_dims": morph_result["n_dims"],
                "morph_mean_rho": morph_result["mean_rho"],
                "morph_mean_p": morph_result["mean_p"],
                "morph_std_rho": morph_result["std_rho"],
                "morph_std_p": morph_result["std_p"],
                "morph_mean_std_joint_rho": morph_result["mean_std_joint_rho"],
                "morph_mean_std_joint_perm_p": morph_result["mean_std_joint_perm_p"],
                "n_permutations": morph_result["mean_std_joint_n_permutations"],
            })

    return pd.DataFrame(rows)
```

```python
from pathlib import Path

import pandas as pd
from typing import Any, Dict, Sequence

from random_projection import DEFAULT_SEEDS
# , MODEL_CONFIGS
from spearman import (
    align_baseline_and_seed_on_pca_dim,
    build_joint_mean_zscore_vectors,
    build_mean_vectors,
    compute_spearman_stats,
    load_full_morphology_summary,
    load_full_script_summary,
    standardize_morphology_summary_df,
    standardize_script_summary_df,
)
```

```python
MODEL_CONFIGS: Dict[str, Dict[str, Any]] = {
    "mistral": {
        "model_name": "mistralai/Mistral-7B-v0.1",
        "tokenizer_path": "mistralai/Mistral-7B-v0.1/tokenizer",
        "candidate_dims": [5, 142, 997, 2084, 3031, 3459, 3938, 4088, 4096],
    },
    "mixtral": {
        "model_name": "mistralai/Mixtral-8x7B-v0.1",
        "tokenizer_path": "mistralai/Mixtral-8x7B-v0.1/tokenizer",
        "candidate_dims": [8, 158, 1111, 2156, 3052, 3457, 3957, 4093, 4096],
    },
    "gpt-oss": {
        "model_name": "gpt-oss",
        "tokenizer_path": "gpt-oss/tokenizer",
        "candidate_dims": [6, 182, 466, 739, 1591, 2264, 2532, 2868, 2880],
    },
}
```

```python
def compute_random_projection_rho(
    baseline_morph,
    control_morph,
    baseline_script,
    control_script,
):
    baseline_morph_std = standardize_morphology_summary_df(baseline_morph)
    control_morph_std = standardize_morphology_summary_df(control_morph)
    baseline_script_std = standardize_script_summary_df(baseline_script)
    control_script_std = standardize_script_summary_df(control_script)

    morph_aligned = align_baseline_and_seed_on_pca_dim(
        baseline_morph_std,
        control_morph_std,
    )
    script_aligned = align_baseline_and_seed_on_pca_dim(
        baseline_script_std,
        control_script_std,
    )
    morph_baseline, morph_control = build_mean_vectors(morph_aligned)
    script_baseline, script_control = build_mean_vectors(script_aligned)
    joint_baseline, joint_control, _, _ = build_joint_mean_zscore_vectors(
        baseline_morph_std,
        control_morph_std,
        baseline_script_std,
        control_script_std,
    )
    return {
        "n_dims_morph": int(len(morph_aligned)),
        "n_dims_script": int(len(script_aligned)),
        "morph_mean_rho": compute_spearman_stats(
            morph_baseline, morph_control
        ).rho,
        "script_mean_rho": compute_spearman_stats(
            script_baseline, script_control
        ).rho,
        "joint_mean_rho": compute_spearman_stats(
            joint_baseline, joint_control
        ).rho,
    }


random_projection_spearman_frames = []
for model_key in ["mistral", "mixtral", "gpt-oss"]:
    # 
    model_config = MODEL_CONFIGS[model_key]
    model_name = model_config["model_name"]
    model_dir = Path("comp") / model_name / "output_proj"
    rp_dir = model_dir / "random_projection"
    baseline_morph = load_full_morphology_summary(
        out_root="comp", model_name=model_name, space_name="output_proj"
    )
    baseline_script = load_full_script_summary(
        out_root="comp", model_name=model_name, space_name="output_proj"
    )
    rp_morph_all = pd.read_csv(rp_dir / "kondrak_global_summary_all_seeds.csv")
    rp_script_all = pd.read_csv(rp_dir / "script_entropy_summary_all_seeds.csv")
    rp_morph_mean = pd.read_csv(rp_dir / "kondrak_global_summary_mean.csv")
    rp_script_mean = pd.read_csv(rp_dir / "script_entropy_summary_mean.csv")

    model_rows = [{
        "model_name": model_name,
        "result_type": "mean_curve",
        "rp_seed": pd.NA,
        **compute_random_projection_rho(
            baseline_morph, rp_morph_mean, baseline_script, rp_script_mean
        ),
    }]
    for rp_seed in DEFAULT_SEEDS:
        seed_morph = rp_morph_all.loc[
            rp_morph_all["rp_seed"].astype(int) == int(rp_seed)
        ].copy()
        seed_script = rp_script_all.loc[
            rp_script_all["rp_seed"].astype(int) == int(rp_seed)
        ].copy()
        model_rows.append({
            "model_name": model_name,
            "result_type": "seed",
            "rp_seed": int(rp_seed),
            **compute_random_projection_rho(
                baseline_morph, seed_morph, baseline_script, seed_script
            ),
        })

    model_spearman_df = pd.DataFrame(model_rows)
    model_spearman_df.to_csv(
        rp_dir / "random_projection_spearman_summary.csv",
        index=False,
    )
    random_projection_spearman_frames.append(model_spearman_df)

random_projection_spearman_df = pd.concat(
    random_projection_spearman_frames,
    axis=0,
    ignore_index=True,
)
random_projection_spearman_df.to_csv(
    Path("comp") / "random_projection_spearman_summary_all_models.csv",
    index=False,
)
random_projection_spearman_df
```

```python
def compute_random_projection_rho_and_pvalue(
    baseline_morph,
    control_morph,
    baseline_script,
    control_script,
):
    baseline_morph_std = standardize_morphology_summary_df(baseline_morph)
    control_morph_std = standardize_morphology_summary_df(control_morph)
    baseline_script_std = standardize_script_summary_df(baseline_script)
    control_script_std = standardize_script_summary_df(control_script)

    morph_aligned = align_baseline_and_seed_on_pca_dim(
        baseline_morph_std,
        control_morph_std,
    )
    script_aligned = align_baseline_and_seed_on_pca_dim(
        baseline_script_std,
        control_script_std,
    )
    morph_baseline, morph_control = build_mean_vectors(morph_aligned)
    script_baseline, script_control = build_mean_vectors(script_aligned)
    joint_baseline, joint_control, _, _ = build_joint_mean_zscore_vectors(
        baseline_morph_std,
        control_morph_std,
        baseline_script_std,
        control_script_std,
    )
    morph_stats = compute_spearman_stats(morph_baseline, morph_control)
    script_stats = compute_spearman_stats(script_baseline, script_control)
    joint_stats = compute_spearman_stats(joint_baseline, joint_control)
    return {
        "n_dims_morph": int(len(morph_aligned)),
        "n_dims_script": int(len(script_aligned)),
        "morph_mean_rho": morph_stats.rho,
        "morph_mean_pvalue": morph_stats.pvalue,
        "script_mean_rho": script_stats.rho,
        "script_mean_pvalue": script_stats.pvalue,
        "joint_mean_rho": joint_stats.rho,
        "joint_mean_pvalue": joint_stats.pvalue,
    }


random_projection_spearman_with_pvalue_frames = []
for model_key, model_config in MODEL_CONFIGS.items():
    model_name = model_config["model_name"]
    model_dir = Path("comp") / model_name / "output_proj"
    rp_dir = model_dir / "random_projection"
    baseline_morph = load_full_morphology_summary(
        out_root="comp", model_name=model_name, space_name="output_proj"
    )
    baseline_script = load_full_script_summary(
        out_root="comp", model_name=model_name, space_name="output_proj"
    )
    rp_morph_all = pd.read_csv(rp_dir / "kondrak_global_summary_all_seeds.csv")
    rp_script_all = pd.read_csv(rp_dir / "script_entropy_summary_all_seeds.csv")
    rp_morph_mean = pd.read_csv(rp_dir / "kondrak_global_summary_mean.csv")
    rp_script_mean = pd.read_csv(rp_dir / "script_entropy_summary_mean.csv")

    model_rows = [{
        "model_name": model_name,
        "result_type": "mean_curve",
        "rp_seed": pd.NA,
        **compute_random_projection_rho_and_pvalue(
            baseline_morph, rp_morph_mean, baseline_script, rp_script_mean
        ),
    }]
    for rp_seed in DEFAULT_SEEDS:
        seed_morph = rp_morph_all.loc[
            rp_morph_all["rp_seed"].astype(int) == int(rp_seed)
        ].copy()
        seed_script = rp_script_all.loc[
            rp_script_all["rp_seed"].astype(int) == int(rp_seed)
        ].copy()
        model_rows.append({
            "model_name": model_name,
            "result_type": "seed",
            "rp_seed": int(rp_seed),
            **compute_random_projection_rho_and_pvalue(
                baseline_morph, seed_morph, baseline_script, seed_script
            ),
        })

    model_spearman_with_pvalue_df = pd.DataFrame(model_rows)
    model_spearman_with_pvalue_df.to_csv(
        rp_dir / "random_projection_spearman_summary_with_pvalues.csv",
        index=False,
    )
    random_projection_spearman_with_pvalue_frames.append(
        model_spearman_with_pvalue_df
    )

random_projection_spearman_with_pvalue_df = pd.concat(
    random_projection_spearman_with_pvalue_frames,
    axis=0,
    ignore_index=True,
)
random_projection_spearman_with_pvalue_df.to_csv(
    Path("comp") / "random_projection_spearman_summary_all_models_with_pvalues.csv",
    index=False,
)
random_projection_spearman_with_pvalue_df
```
